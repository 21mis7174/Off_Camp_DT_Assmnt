# Hyderabad Region Filter

This notebook loads the cleaned DSIR company table, filters for a broad Hyderabad region using address keywords, and exports the shortlist for Part A research.

The code is stdlib-only, so it runs in any active Python kernel. Use the existing project virtual environment at `/home/uday/Documents/offCampus_placement_assignments/primetrade.ai/.venv` if you want the project setup.

```python
from pathlib import Path
import csv
import re
import sys

BASE_DIR = Path('/home/uday/Documents/offCampus_placement_assignments/DeepThought/round1')
INPUT_CSV = BASE_DIR / 'output' / 'supporting_csv' / 'dsir_recognized_rd_centres.csv'
OUTPUT_CSV = BASE_DIR / 'output' / 'supporting_csv' / 'companies_hyderabad_broad.csv'

BROAD_HYDERABAD_TERMS = [
    'hyderabad',
    'secunderabad',
    'medchal',
    'ranga reddy',
    'rangareddy',
    'sangareddy',
    'patancheru',
    'jeedimetla',
    'kukatpally',
    'gachibowli',
    'kondapur',
    'uppal',
    'shamirpet',
    'cyberabad',
    'madhapur',
    'miyapur',
    'hitech city',
    'quthbullapur',
    'qutbullapur',
    'shamshabad',
    'adibatla',
    'shankarpally',
    'kompally',
    'bachupally',
    'pocharam',
    'gundlapochampally'
]

print(f'python={sys.executable}')
print(f'input={INPUT_CSV}')
print(f'output={OUTPUT_CSV}')
```

```python
def normalize_text(value: str) -> str:
    text = (value or '').strip().lower()
    text = re.sub(r'\s+', ' ', text)
    return text

def matches_hyderabad_region(company_name: str, company_address: str) -> bool:
    haystack = normalize_text(f'{company_name} {company_address}')
    return any(term in haystack for term in BROAD_HYDERABAD_TERMS)

rows = []
with INPUT_CSV.open('r', encoding='utf-8', newline='') as handle:
    reader = csv.DictReader(handle)
    for row in reader:
        if matches_hyderabad_region(row.get('Company Name', ''), row.get('Company Address', '')):
            rows.append(row)

print(f'total_rows={len(rows)}')
rows[:3]
```

```python
fieldnames = []
seen = set()
for row in rows:
    key = (normalize_text(row.get('Company Name', '')), normalize_text(row.get('Company Address', '')))
    if key in seen:
        continue
    seen.add(key)
    if not fieldnames:
        fieldnames = list(row.keys())

deduped_rows = []
seen = set()
for row in rows:
    key = (normalize_text(row.get('Company Name', '')), normalize_text(row.get('Company Address', '')))
    if key in seen:
        continue
    seen.add(key)
    deduped_rows.append(row)

with OUTPUT_CSV.open('w', encoding='utf-8', newline='') as handle:
    writer = csv.DictWriter(handle, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(deduped_rows)

print(f'written={OUTPUT_CSV}')
print(f'result_rows={len(deduped_rows)}')
deduped_rows[:10]
```
